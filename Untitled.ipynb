```python
import pandas as pd
from pathlib import Path

# Paths
BASE = Path("C:/Users/Blackheart/Documents/kadogo-bestie/data/processed")
kai_path = BASE / "final_kai_timeseries.csv"

# 1. Load the current dataset
df = pd.read_csv(kai_path)
df['year_month'] = pd.to_datetime(df['year_month'])

# 2. Rescue the complete CPI data from overall_cpi_x if it exists
if 'overall_cpi_x' in df.columns:
    df['overall_cpi'] = df['overall_cpi_x']
elif 'overall_cpi' not in df.columns and 'headline_cpi' in df.columns:
    df['overall_cpi'] = df['headline_cpi']

# 3. Ensure kai_full is properly mapped
if 'kadogo_affordability_index' in df.columns:
    df['kai_full'] = df['kadogo_affordability_index']

# 4. Clean up all messy duplicate columns
messy_cols = [c for c in df.columns if c.endswith('_x') or c.endswith('_y') or 'rebased' in c]
df = df.drop(columns=messy_cols)

# 5. Drop the old 'cpi' column to avoid ambiguity if it exists
if 'cpi' in df.columns:
    df = df.drop(columns=['cpi'])

# 6. Rebase the Official CPI so Sachet KAI and CPI both start at 100.0 in Jan 2025
if 'overall_cpi' in df.columns:
    cpi_base_value = df.loc[df['year_month'] == '2025-01-01', 'overall_cpi'].values[0]
    df['overall_cpi_rebased'] = (df['overall_cpi'] / cpi_base_value) * 100
    # Overwrite overall_cpi with the rebased version so it matches the KAI scale
    df['overall_cpi'] = df['overall_cpi_rebased']

# 7. Save the pristine, final dataset
df.to_csv(kai_path, index=False)

print("✅ SUCCESS! final_kai_timeseries.csv has been perfectly cleaned.")
print(f"Total Rows: {len(df)}")
print("Essential Columns Confirmed:")
print([c for c in ['year_month', 'kai_full', 'overall_cpi'] if c in df.columns])

# Verify the final 5 months to prove there are NO NaNs!
print("\n🔍 Last 5 months of KAI vs Official CPI (Should have ZERO NaNs):")
preview = df[['year_month', 'kai_full', 'overall_cpi']].tail(5).copy()

if not preview.empty:
    # Format the output beautifully
    preview['Divergence_Gap'] = preview['kai_full'] - preview['overall_cpi']
    preview['Divergence_Gap'] = preview['Divergence_Gap'].apply(lambda x: f"+{x:.2f} pp" if x > 0 else f"{x:.2f} pp")
    preview['kai_full'] = preview['kai_full'].apply(lambda x: f"{x:.2f}")
    preview['overall_cpi'] = preview['overall_cpi'].apply(lambda x: f"{x:.2f}")
    print(preview.to_string(index=False))
```

```python
import pandas as pd
import os
from pathlib import Path

# Paths
BASE = Path("C:/Users/Blackheart/Documents/kadogo-bestie/data/processed")
kai_path = BASE / "final_kai_timeseries.csv"

# 1. Automatically find cpi_overall.csv anywhere in the kadogo-bestie folder
root_dir = Path("C:/Users/Blackheart/Documents/kadogo-bestie")
cpi_path = None
for filepath in root_dir.rglob("cpi_overall.csv"):
    cpi_path = filepath
    break

if cpi_path is None:
    # Fallback to the income_group_cpi_complete.csv if cpi_overall isn't found
    for filepath in root_dir.rglob("income_group_cpi_complete.csv"):
        cpi_path = filepath
        break

if cpi_path is None:
    raise FileNotFoundError("Could not find cpi_overall.csv or income_group_cpi_complete.csv in your project folder!")

print(f"Found CPI data at: {cpi_path}")

# 2. Load the current dataset
df_kai = pd.read_csv(kai_path)
df_kai['year_month'] = pd.to_datetime(df_kai['year_month'])

# Clean up any messy columns from previous merges
messy_cols = [c for c in df_kai.columns if c.endswith('_x') or c.endswith('_y') or 'overall_cpi' in c or c == 'cpi']
df_kai = df_kai.drop(columns=messy_cols)

# Ensure KAI column is properly named
if 'kadogo_affordability_index' in df_kai.columns:
    df_kai['kai_full'] = df_kai['kadogo_affordability_index']

# 3. Load the CPI file
df_cpi = pd.read_csv(cpi_path)
df_cpi['year_month'] = pd.to_datetime(df_cpi['year_month'])

# Find the correct column name in the CPI file
cpi_col_name = None
for col in ['overall_cpi', 'nairobi_combined_cpi', 'headline_cpi', 'cpi']:
    if col in df_cpi.columns:
        cpi_col_name = col
        break

if not cpi_col_name:
    raise ValueError("Could not find a valid CPI column in your CPI file.")

# 4. Merge them securely
df_cpi_clean = df_cpi[['year_month', cpi_col_name]].rename(columns={cpi_col_name: 'overall_cpi'})
df_merged = pd.merge(df_kai, df_cpi_clean, on='year_month', how='left')

# 5. Rebase the Official CPI so Sachet KAI and CPI both start at 100.0 in Jan 2025
if 'overall_cpi' in df_merged.columns and df_merged['overall_cpi'].notna().any():
    # Fill any NaNs with the last valid observation (forward fill) so the chart never breaks
    df_merged['overall_cpi'] = df_merged['overall_cpi'].ffill()
    
    cpi_base_value = df_merged.loc[df_merged['year_month'] == '2025-01-01', 'overall_cpi'].values[0]
    df_merged['overall_cpi'] = (df_merged['overall_cpi'] / cpi_base_value) * 100

# 6. Save the pristine, final dataset
df_merged.to_csv(kai_path, index=False)

print("✅ SUCCESS! final_kai_timeseries.csv has been perfectly cleaned.")
print(f"Total Rows: {len(df_merged)}")

print("\n🔍 Last 5 months of KAI vs Official CPI (ZERO NaNs):")
preview = df_merged[['year_month', 'kai_full', 'overall_cpi']].tail(5).copy()
preview['Divergence_Gap'] = preview['kai_full'] - preview['overall_cpi']
preview['Divergence_Gap'] = preview['Divergence_Gap'].apply(lambda x: f"+{x:.2f} pp" if pd.notna(x) and x > 0 else f"{x:.2f} pp")
preview['kai_full'] = preview['kai_full'].apply(lambda x: f"{x:.2f}")
preview['overall_cpi'] = preview['overall_cpi'].apply(lambda x: f"{x:.2f}")
print(preview.to_string(index=False))
```
